```python
from matplotlib import pyplot as plt
```

```python
x =[1,2,3,4]
y= [2,3,4,5]

plt.plot(x,y)
c =["r","b","g","y"]

plt.bar(x,y,color=c,width=0.5,edgecolor="k")
plt.scatter(x,y)
plt.xlabel("ages_x population")
plt.ylabel("second")
plt.title("It's a title")
plt.show()
```

```python
import matplotlib.pyplot as plt
x=["python","c","c++","java"]
y = [85,60,70,82]
plt.xlabel("language",fontsize=15)
plt.ylabel("numbers by popularity")
plt.title("popularity of programming language")
c = ["y","g","m","y"]
plt.bar(x,y,color=c,edgecolor="black",linewidth=1,linestyle=":",alpha=0.5)
# linewidth is used for draw edge line
#alpha is used to blur the bar
#linestyle is used to draw edge line in anyform
#label is used to write for what reson this graph is
plt.show()
```

```python
import matplotlib.pyplot as plt
x=["python","c","c++","java"]
y = [85,60,70,82]
plt.xlabel("language",fontsize=15)
plt.ylabel("numbers by popularity")
plt.title("popularity of programming language")
c = ["y","g","m","y"]
plt.bar(x,y,color=c,edgecolor="black",label="Mahadev")
plt.legend()
plt.show()

```
